### Análise: Pergunta 1 — Taxa de inadimplência geral e por perfil

```python
%sql
SELECT
  ROUND(AVG(TARGET) * 100, 2) AS taxa_inadimplencia_geral_pct
FROM credit_risk_pipeline.gold.perfil_risco_cliente
```

```python
%sql
SELECT
  NAME_CONTRACT_TYPE,
  COUNT(*) AS qtd_clientes,
  ROUND(AVG(TARGET) * 100, 2) AS taxa_inadimplencia_pct
FROM credit_risk_pipeline.gold.perfil_risco_cliente
GROUP BY NAME_CONTRACT_TYPE
ORDER BY taxa_inadimplencia_pct DESC
```

```python
%sql
SELECT
  NAME_EDUCATION_TYPE,
  COUNT(*) AS qtd_clientes,
  ROUND(AVG(TARGET) * 100, 2) AS taxa_inadimplencia_pct
FROM credit_risk_pipeline.gold.perfil_risco_cliente
GROUP BY NAME_EDUCATION_TYPE
ORDER BY taxa_inadimplencia_pct DESC
```

### Análise: Pergunta 2 — Atraso no bureau x inadimplência atual

```python
%sql
SELECT
  CASE
    WHEN atraso_medio_dias_bureau IS NULL THEN 'Sem histórico no bureau'
    WHEN atraso_medio_dias_bureau > 0 THEN 'Com atraso no bureau'
    ELSE 'Sem atraso no bureau'
  END AS grupo_bureau,
  COUNT(*) AS qtd_clientes,
  ROUND(AVG(TARGET) * 100, 2) AS taxa_inadimplencia_pct
FROM credit_risk_pipeline.gold.perfil_risco_cliente
GROUP BY grupo_bureau
ORDER BY taxa_inadimplencia_pct DESC
```

### Análise: Pergunta 3 — Comportamento de pagamento x risco atual

```python
%sql
SELECT
  CASE
    WHEN atraso_medio_dias_pagamento IS NULL THEN 'Sem parcelas anteriores registradas'
    WHEN pct_parcelas_atrasadas > 0.3 THEN 'Atrasou mais de 30% das parcelas'
    WHEN pct_parcelas_atrasadas > 0 THEN 'Atrasou até 30% das parcelas'
    ELSE 'Nunca atrasou parcela'
  END AS grupo_pagamento,
  COUNT(*) AS qtd_clientes,
  ROUND(AVG(TARGET) * 100, 2) AS taxa_inadimplencia_pct
FROM credit_risk_pipeline.gold.perfil_risco_cliente
GROUP BY grupo_pagamento
ORDER BY taxa_inadimplencia_pct DESC
```

### Análise: Pergunta 4 — Exposição financeira por segmento

```python
%sql
SELECT
  faixa_etaria,
  faixa_renda,
  faixa_valor_credito,
  qtd_clientes,
  taxa_inadimplencia,
  exposicao_total_credito,
  exposicao_em_risco,
  ROUND(exposicao_em_risco / exposicao_total_credito * 100, 2) AS pct_exposicao_em_risco
FROM credit_risk_pipeline.gold.indicadores_risco
ORDER BY exposicao_em_risco DESC
LIMIT 10
```
