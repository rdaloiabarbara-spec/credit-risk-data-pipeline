### Bronze: Ingestão dos dados brutos
Lê os 4 arquivos CSV originais do volume `raw_files` e grava cada um como tabela Delta no schema `bronze`, sem nenhuma alteração de valor. Adiciona apenas metadados de controle (data de ingestão e arquivo de origem).

```python
from pyspark.sql import functions as F

def ingerir_bronze(nome_arquivo, nome_tabela):
    caminho = f"/Volumes/credit_risk_pipeline/bronze/raw_files/{nome_arquivo}"

    df = (
        spark.read
        .option("header", True)
        .option("inferSchema", True)
        .csv(caminho)
        .withColumn("_ingestion_timestamp", F.current_timestamp())
        .withColumn("_source_file", F.lit(nome_arquivo))
    )

    df.write.mode("overwrite").saveAsTable(f"credit_risk_pipeline.bronze.{nome_tabela}")
    print(f"Tabela {nome_tabela} criada com {df.count()} linhas.")
```

```python
ingerir_bronze("application_train.csv", "application")
ingerir_bronze("bureau.csv", "bureau")
ingerir_bronze("previous_application.csv", "previous_application")
ingerir_bronze("installments_payments.csv", "installments_payments")
```
